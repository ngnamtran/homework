````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def middlest_element(values):
    """Returns the middle elelment of a list.
    If the list is empty, return None. 
    If the list length is odd, return the element art exact middle index.
    If the list length is even, return the element left of center
    """
    if len(values) == 0:
        return None
    elif len(values) %2 == 0:
        return values[len(values)//2 - 1]
    else:
        return values[len(values)//2]

````

**Problem 7**

**Concept:** *Accumulator pattern with numeric result, list iteration.*

**Task:**

Write a function `variance(values)` that returns the population variance of a list of numbers using this formula:

$$
\text{variance} = \frac{1}{n} \sum_{i=1}^{n} (x_i - \bar{x})^2
$$

Rules:

* Must use loops + accumulator (no `sum()`).

* No imports.

* Must call your `mean(values)` function to get the mean.

* Then run the provided test case and print in the same format.

**Sample Output:**

```
values: [6, 3, 7, 9, 5]
mean: 6.0
variance: 4.0
```


````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def variance(values):
    """Returns tthe population variance of a list of numbers, using the previous function mean(values)"""
    #define an accumulator
    results= 0
    #use a loop to iterate
    for i in values:
        sub_squared= (i-mean(values))**2
        results += sub_squared
    # after loop, use accumulate value
    variance = results / len(values)
    return variance
````

**Problem 8**

**Concept:** *Accumulator pattern with numeric result, list iteration, index-based loop.*

**Task:**

Write a function `covariance(xs, ys)` that computes the population covariance between two series:

$$
\text{covariance} = \frac{1}{n} \sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})
$$

Rules:

* If `xs` and `ys` are not the same length, print:
    
    `xs and ys must be the same length`
    
    and return `0`.

* Must use an index-based loop.

* Must use your `mean()` function.

* Then run the given test case and print exactly in the same format.

**Sample Output:**

```
x values: [6, 3, 7, 9, 5]
y values: [4, 1, 7, 5, 0]
covariance: 3.8
```


````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def covariance(xs,ys):
    """Compute the population covariance between two series.
        Two parameters must be the same length"""
    if len(xs) != len(ys):
        print('xs and ys must be the same length')
        return 0
    else:
        #define accumulator
        sum_multiply = 0
        #use loop to iterate
        for i in range(len(xs)):
            sum_substract_xs = xs[i]-mean(xs)
            sum_substract_ys = ys[i]-mean(ys)
            #use accumulate
            sum_multiply += sum_substract_xs*sum_substract_ys
        #use accumulative result
        return sum_multiply/len(xs)
            
````

**Problem 12**

**Concept:** *Accumulator pattern to process a string, produce a string result.*

**Task:**

Write a function `cipher(s, n)` that applies a Caesar cipher shift to a plaintext string.

Rules:

* Assume lowercase letters only for shifting (`a` to `z`).

* Do not shift non-letters (spaces stay spaces, punctuation stays punctuation).

* Must wrap around the alphabet if shifting past `z` or before `a`.

* Must use an accumulator string.

* Then run the exact test cases shown and print in the same format.

**Sample Output:**

```
plaintext = abc, ciphered = def
plaintext = abc xyz, ciphered = def abc
```


````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def cipher(s,n):
    """Applies a Caesar cipher sift to a plaintext string
        Assunging lower cases letters only and not shift non-letters"""
    #define accumulate
    ciphered = ''
    #use loop
    for i in s:
        #assuming lower case
        if i>= 'a' and i<= 'z':
            code_i= ord(i)+n
            #if shifing past z:
            if code_i> ord('z'):
                ciphered += chr(code_i-26)
            #if shifting before'a'
            elif code_i<ord('a') :
                ciphered +=chr(code_i+26)
            else:
                ciphered +=chr(code_i)
        #not shift any non-letters
        else:
            ciphered += i
    
    return ciphered
    
````

Test case for problem 12.
